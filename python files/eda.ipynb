**Correlation Matrix**

```python
correlation_matrix = df[numeric_columns].corr()

correlation_matrix
```

**Correlation ranking**

```python
corr_target = df.corr(numeric_only=True)["Mortgage_Approvals"].sort_values(ascending=False)
print(corr_target)
```

**Correlation Heatmap**

```python
plt.figure(figsize=(12,8))
sns.heatmap(correlation_matrix,
            annot=True,
            cmap="coolwarm",
            fmt=".2f",
            linewidths=0.5)

plt.title("Correlation Matrix of Macroeconomic and Retail Banking Variables")
plt.show()
```
